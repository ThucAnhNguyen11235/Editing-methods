```python
import os, sys, gc, torch, time, json
from pathlib import Path

PYDEPS = "/kaggle/working/pydeps"
REPO = "/kaggle/working/EasyEdit"

sys.path.insert(0, PYDEPS)
sys.path.insert(0, REPO)

os.environ["PYTHONPATH"] = f"{PYDEPS}:{REPO}:" + os.environ.get("PYTHONPATH", "")
os.environ["TOKENIZERS_PARALLELISM"] = "false"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

from easyeditor import EditTrainer, MENDTrainingHparams, ZsreDataset

def clear_cuda():
    gc.collect()
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

def get_gpu_snapshot():
    if not torch.cuda.is_available():
        return {}

    return {
        "allocated_gb": torch.cuda.memory_allocated() / 1e9,
        "reserved_gb": torch.cuda.memory_reserved() / 1e9,
        "max_allocated_gb": torch.cuda.max_memory_allocated() / 1e9,
        "max_reserved_gb": torch.cuda.max_memory_reserved() / 1e9,
    }

def train_one_sweep(n_train: int):
    clear_cuda()
    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()

    hparams_path = f"/kaggle/working/EasyEdit/hparams/TRAINING/MEND/llama32_1b_zsre_n{n_train}.yaml"
    train_path = f"/kaggle/working/EasyEdit/data/zsre_sweeps/train_{n_train}.json"
    eval_path = "/kaggle/working/EasyEdit/data/zsre_sweeps/eval_100.json"

    print("\n" + "=" * 80)
    print("TRAINING:", n_train)
    print("hparams:", hparams_path)
    print("train:", train_path)
    print("eval:", eval_path)
    print("=" * 80)

    for p in [hparams_path, train_path, eval_path]:
        p = Path(p)
        print(p, "exists:", p.exists(), "size MB:", round(p.stat().st_size / 1e6, 3) if p.exists() else None)

    training_hparams = MENDTrainingHparams.from_hparams(hparams_path)

    train_ds = ZsreDataset(train_path, config=training_hparams)
    eval_ds = ZsreDataset(eval_path, config=training_hparams)

    print("train size:", len(train_ds))
    print("eval size:", len(eval_ds))
    print("results_dir:", training_hparams.results_dir)
    print("max_iters:", training_hparams.max_iters)
    print("batch_size:", training_hparams.batch_size)
    print("accumulate_bs:", training_hparams.accumulate_bs)

    trainer = EditTrainer(
        config=training_hparams,
        train_set=train_ds,
        val_set=eval_ds,
    )

    print("save_path:", trainer.save_path)

    result_dir = Path(training_hparams.results_dir)
    result_dir.mkdir(parents=True, exist_ok=True)

    runtime_path = result_dir / "training_runtime.json"

    start_wall = time.perf_counter()
    start_time_readable = time.strftime("%Y-%m-%d %H:%M:%S")

    status = "success"
    error = None

    try:
        trainer.run()

    except Exception as e:
        status = "error"
        error = repr(e)
        raise

    finally:
        end_wall = time.perf_counter()
        end_time_readable = time.strftime("%Y-%m-%d %H:%M:%S")
        elapsed_seconds = end_wall - start_wall

        runtime_record = {
            "n_train": n_train,
            "status": status,
            "error": error,
            "start_time": start_time_readable,
            "end_time": end_time_readable,
            "elapsed_seconds": elapsed_seconds,
            "elapsed_minutes": elapsed_seconds / 60,
            "elapsed_hours": elapsed_seconds / 3600,
            "train_size": len(train_ds),
            "eval_size": len(eval_ds),
            "max_iters": getattr(training_hparams, "max_iters", None),
            "batch_size": getattr(training_hparams, "batch_size", None),
            "accumulate_bs": getattr(training_hparams, "accumulate_bs", None),
            "effective_batch_size": getattr(training_hparams, "batch_size", 1) * getattr(training_hparams, "accumulate_bs", 1),
            "results_dir": str(result_dir),
            "save_path": str(trainer.save_path),
            "gpu": get_gpu_snapshot(),
        }

        with open(runtime_path, "w") as f:
            json.dump(runtime_record, f, indent=2)

        print("\nTraining runtime record:")
        print(json.dumps(runtime_record, indent=2))
        print("Saved:", runtime_path)

        del trainer
        del train_ds
        del eval_ds
        del training_hparams
        clear_cuda()

    return result_dir
```

```python
train_results = {}

for n in [1000, 2000, 5000]:
    train_results[n] = str(train_one_sweep(n))
    print("finished:", n, train_results[n])
    clear_cuda()
```

```python
import yaml
from pathlib import Path

def write_eval_yaml(n_train: int):
    train_yaml = Path(f"/kaggle/working/EasyEdit/hparams/TRAINING/MEND/llama32_1b_zsre_n{n_train}.yaml")
    eval_yaml = Path(f"/kaggle/working/EasyEdit/hparams/MEND/llama32_1b_zsre_n{n_train}.yaml")
    eval_yaml.parent.mkdir(parents=True, exist_ok=True)

    ckpt = Path(f"/kaggle/working/mend_sweeps/llama32_1b_zsre_n{n_train}/models/MEND/Llama-3.2-1B-Instruct")

    cfg = yaml.safe_load(train_yaml.read_text())

    cfg["alg_name"] = "MEND"
    cfg["lr_scale"] = 1.0
    cfg["archive"] = str(ckpt)
    cfg["eval_only"] = True
    cfg["save"] = False
    cfg["final_eval"] = False
    cfg["results_dir"] = f"/kaggle/working/mend_sweeps/llama32_1b_zsre_n{n_train}_eval"

    eval_yaml.write_text(yaml.safe_dump(cfg, sort_keys=False))

    print("eval yaml:", eval_yaml)
    print("archive exists:", ckpt.exists(), ckpt)

    return eval_yaml

for n in [1000, 2000, 5000]:
    write_eval_yaml(n)
```

```python
import json, gc, torch, numpy as np, time
from pathlib import Path

from easyeditor import MENDHyperParams, BaseEditor

def scalarize(x):
    if isinstance(x, np.generic):
        return x.item()
    if isinstance(x, list):
        return [scalarize(v) for v in x]
    if isinstance(x, dict):
        return {k: scalarize(v) for k, v in x.items()}
    return x

def first_float(v):
    if isinstance(v, list):
        return float(v[0]) if len(v) else None
    if v is None:
        return None
    return float(v)

def summarize(metrics_list):
    reliability, generalization, locality, portability, runtime = [], [], [], [], []

    for m in metrics_list:
        post = m.get("post", {})

        if "rewrite_acc" in post:
            val = first_float(post["rewrite_acc"])
            if val is not None:
                reliability.append(val)

        if "rephrase_acc" in post:
            val = first_float(post["rephrase_acc"])
            if val is not None:
                generalization.append(val)

        loc = post.get("locality", {})
        if "neighborhood_acc" in loc:
            val = first_float(loc["neighborhood_acc"])
            if val is not None:
                locality.append(val)

        port = post.get("portability", {})
        if "one_hop_acc" in port:
            val = first_float(port["one_hop_acc"])
            if val is not None:
                portability.append(val)

        if "runtime_seconds_wall" in m:
            runtime.append(float(m["runtime_seconds_wall"]))

    return {
        "n": len(metrics_list),
        "reliability_rewrite_acc": sum(reliability) / len(reliability) if reliability else None,
        "generalization_rephrase_acc": sum(generalization) / len(generalization) if generalization else None,
        "locality_neighborhood_acc": sum(locality) / len(locality) if locality else None,
        "portability_one_hop_acc": sum(portability) / len(portability) if portability else None,
        "avg_runtime_seconds_per_edit": sum(runtime) / len(runtime) if runtime else None,
        "total_runtime_seconds": sum(runtime) if runtime else None,
    }

def eval_one_sweep(n_train: int, n_eval: int = 100):
    clear_cuda()

    eval_yaml = f"/kaggle/working/EasyEdit/hparams/MEND/llama32_1b_zsre_n{n_train}.yaml"
    eval_path = "/kaggle/working/EasyEdit/data/zsre_sweeps/eval_100.json"

    out_path = Path(f"/kaggle/working/mend_sweeps/llama32_1b_zsre_n{n_train}_eval_metrics_{n_eval}.jsonl")
    json_out_path = Path(f"/kaggle/working/mend_sweeps/llama32_1b_zsre_n{n_train}_eval_metrics_{n_eval}.json")

    eval_items = json.load(open(eval_path))[:n_eval]

    hparams = MENDHyperParams.from_hparams(eval_yaml)
    editor = BaseEditor.from_hparams(hparams)

    all_metrics = []
    out_path.write_text("")

    for i, x in enumerate(eval_items):
        print(f"\n=== [{n_train}] Eval item {i+1}/{n_eval} ===")

        ground_truth = (
            x["answers"][0]
            if isinstance(x.get("answers"), list) and len(x["answers"])
            else x.get("pred", "")
        )

        rephrase_prompt = x.get("rephrase", None)

        locality_inputs = {
            "neighborhood": {
                "prompt": [x["loc"]],
                "ground_truth": [x["loc_ans"]],
            }
        }

        portability_inputs = {}
        if x.get("portability"):
            portability_inputs = {
                "one_hop": {
                    "prompt": [x["portability"]["New Question"]],
                    "ground_truth": [x["portability"]["New Answer"]],
                }
            }

        try:
            start = time.perf_counter()

            edit_kwargs = dict(
                prompts=[x["src"]],
                subject=[x["subject"]],
                ground_truth=[ground_truth],
                target_new=[x["alt"]],
                locality_inputs=locality_inputs,
                portability_inputs=portability_inputs,
                keep_original_weight=True,
                test_generation=False,
            )

            if rephrase_prompt:
                edit_kwargs["rephrase_prompts"] = [rephrase_prompt]

            metrics, edited_model, _ = editor.edit(**edit_kwargs)

            elapsed = time.perf_counter() - start

            m = scalarize(metrics[0] if isinstance(metrics, list) else metrics)
            m["runtime_seconds_wall"] = elapsed
            all_metrics.append(m)

            with open(out_path, "a") as f:
                f.write(json.dumps(m) + "\n")

            post = m.get("post", {})
            print("post reliability/rewrite:", post.get("rewrite_acc"))
            print("post generalization/rephrase:", post.get("rephrase_acc"))
            print("post locality:", post.get("locality"))
            print("post portability:", post.get("portability"))
            print("runtime seconds:", round(elapsed, 4))

            del edited_model
            gc.collect()
            torch.cuda.empty_cache()
            torch.cuda.ipc_collect()

            if (i + 1) % 10 == 0:
                print("\nRunning summary:", summarize(all_metrics))

        except torch.cuda.OutOfMemoryError:
            print("OOM at item", i)
            gc.collect()
            torch.cuda.empty_cache()
            torch.cuda.ipc_collect()
            break

        except Exception as e:
            print("Error at item", i, repr(e))
            gc.collect()
            torch.cuda.empty_cache()
            torch.cuda.ipc_collect()
            continue

    summary = summarize(all_metrics)

    with open(json_out_path, "w") as f:
        json.dump(
            {
                "n_train": n_train,
                "n_eval": len(all_metrics),
                "summary": summary,
                "metrics": all_metrics,
            },
            f,
            indent=2,
        )

    print("\nFinal summary:", summary)
    print("Saved JSONL:", out_path)
    print("Saved JSON:", json_out_path)

    del editor
    del hparams
    clear_cuda()

    return summary
```

```python
sweep_summaries = {}

for n in [1000, 2000, 5000]:
    sweep_summaries[n] = eval_one_sweep(n, n_eval=100)
    clear_cuda()

sweep_summaries
```
